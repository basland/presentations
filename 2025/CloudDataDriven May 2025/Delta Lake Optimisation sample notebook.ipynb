```python
# create a very messy data by forcing small files (maxRecordsPerFile)
# use this copy to show the effect of OPTIMIZE

df = spark.read.table("0_demo_taxi2024_messy") \
    .repartition(10000)

df.write.format("delta") \
    .option("maxRecordsPerFile", 30000) \
    .mode("overwrite") \
    .saveAsTable("0_demo_taxi2024_messy_optimize_demo")
```

```python
%%sql
--Showcase the OPTIMIZE command, it will create few large files from many small files
OPTIMIZE 0_demo_taxi2024_messy_optimize_demo
```
